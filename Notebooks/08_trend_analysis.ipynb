## Making trend analysis file 

```python
import pandas as pd
import numpy as np 
```

```python
commodity_monthly = pd.read_csv("../data/processed/commodity_monthly_price.csv")
```

```python
commodity_monthly.shape
```

```python
commodity_monthly=commodity_monthly.sort_values(by=["commodity","year","month"])
```

as we have to compare with last months price then we have ot make new column which has previous months price for that we use shift()
shift(1):- move row forward by 1 
shift(-1):- move backward by 1

```python
commodity_monthly["previous_month_price"] = (commodity_monthly.groupby(["commodity","year"])["avg_modal_price"].shift(1))
```

```python
commodity_monthly["price_difference"] = (commodity_monthly["avg_modal_price"] -commodity_monthly["previous_month_price"])
```

```python
commodity_monthly["month_over_month_percentage_change"] = (commodity_monthly["price_difference"] /commodity_monthly["previous_month_price"]) * 100
```

```python
conditions = [
    commodity_monthly["price_difference"] > 0,
    commodity_monthly["price_difference"] < 0,
    commodity_monthly["price_difference"] == 0
]

choices = [
    "Increasing",
    "Decreasing",
    "Stable"
]

commodity_monthly["trend"] = np.select(
    conditions,
    choices,
    default="No Previous Month"
)
```

dataFrame has thousands of rows.
np.select() applies the conditions to the entire column at once, which is much faster and is the standard pandas/NumPy approach.

```python
columns_to_keep = ["commodity","year","month","avg_modal_price","previous_month_price","price_difference","month_over_month_percentage_change","trend"]
```

```python
df=commodity_monthly[columns_to_keep]
```

```python
df.to_csv("../data/processed/trend_analysis.csv",index=False)

print("Trend analysis completed successfully.")
```
