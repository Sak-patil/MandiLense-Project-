```python
import pandas as pd 
import numpy as np
```

### 1.Top Expensive Commodities

```python
commodity_stats = pd.read_csv(
    "../data/processed/commodity_statistics.csv"
)

top_expensive = (
    commodity_stats
    .sort_values(
        by="mean_price",
        ascending=False
    )
)

top_expensive.head(10)
```

Varience and standard deviation are shown in scientific notation 

Certain commodities such as Kankambra and Astera exhibit exceptionally high coefficients of variation (>500%), indicating either highly volatile market prices or the presence of extreme outlier values. These commodities require further investigation before making pricing recommendations

Cheapest Commodities

```python
cheapest = (
    commodity_stats
    .sort_values(
        by="mean_price",
        ascending=True 
    )
)
cheapest.head()
```

### Most Volatile Commodities

```python
volatile=(commodity_stats.sort_values(by="coefficient_of_variation",ascending=False))
volatile.head()
```

As here we can see the coefficient of variation is larger than 100 means outliers present in that commodity

Most Stable Commodities

```python
volatile=(commodity_stats.sort_values(by="coefficient_of_variation",ascending=True))
volatile.head()
```

### Best Selling Season

```python
seasonal_csv=pd.read_csv("../data/processed/seasonal_price_summary.csv")
idx = (seasonal_csv.groupby(["commodity", "year"])["avg_modal_price"].idxmax())
best_seasons=seasonal_csv.loc[idx]
# best_seasons[["commodity","year","season"]]
best_seasons[best_seasons["commodity"]=="Ajwan"][["commodity","year","season"]]
```

Worst Selling Season

```python
idx = (seasonal_csv.groupby(["commodity", "year"])["avg_modal_price"].idxmin())
worst_seasons=seasonal_csv.loc[idx]
worst_seasons[["commodity","season","year","avg_modal_price"]].head()
```

### Highest Price State

```python
state_csv=pd.read_csv("../data/processed/state_commodity_price.csv")
idx=state_csv.groupby("commodity")["avg_modal_price"].idxmax()
high_price_states=state_csv.loc[idx]
high_price_states[["state","commodity","year","avg_modal_price"]].head()
```

### Highest Price Markets

```python
market_csv=pd.read_csv("../data/processed/market_price_summary.csv")
markets=market_csv.groupby("commodity")["avg_modal_price"].idxmax()
high_price_market=market_csv.loc[markets]
high_price_market[["state","commodity","year","avg_modal_price","district","market"]].head()
```

### Fastest Growing Commodities

```python
trend=pd.read_csv("../data/processed/trend_analysis.csv")

positive_trends = trend[trend["month_over_month_percentage_change"] > 0]
high_increase_idx=positive_trends.groupby(["commodity","year"])["month_over_month_percentage_change"].idxmax()
```
